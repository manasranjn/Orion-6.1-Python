## Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Data-set

```python
df = pd.read_csv("student_marks.csv")
df.head()
```

## Calculate Average marks

```python
df["Average"] = df[["Math", "Science", "English", "History"]].mean(axis=1)
df
```

## Find Topper

```python
topper = df.loc[df["Average"].idxmax()]
topper
```

## Find class average

```python
class_avr = np.mean(df['Average'])
class_avr
```

```python
print("Subject wise highest marks")
df[["Math", "Science", "English", "History"]].max()
```

## bar Chart for student average marks

```python
plt.barh(df["Name"], df["Average"])
plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")
```

## Subject Chart

```python
subject_avr = df[["Math", "Science", "English", "History"]].mean()
subject_avr
```

```python
plt.bar(["Math", "Science", "English", "History"], subject_avr, label="Subject Average")
plt.title("Average Marks per subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
```
