```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("sales_data.csv")
df
```

## Total revenue calculate

```python
total_sales = df["Sales"].sum()
total_sales
```

## Total Profit

```python
total_profit = df["Profit"].sum()
total_profit
```

## Best Selling Product

```python
best_product = df.groupby("Product")['Sales'].sum()
best_product
```

```python
plt.bar(best_product.index, best_product.values)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
```

## Profit Chart

```python
profits = df.groupby("Product")["Profit"].sum()

plt.bar(profits.index, profits.values)
plt.title("Product Profits")
plt.xlabel("Products")
plt.ylabel("Sales")
```

```python
plt.plot(df["Date"], df["Sales"])
plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
```

```python
plt.pie(best_product.values, labels=best_product.index)
```
